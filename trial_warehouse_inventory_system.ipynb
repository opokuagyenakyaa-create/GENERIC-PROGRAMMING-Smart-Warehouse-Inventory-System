# Trial: Generic Storage Class
### Case Study Activity 1

```python
from typing import TypeVar, Generic

T = TypeVar('T')

class Storage(Generic[T]):
    def __init__(self):
        self.__item = None

    def store(self, item: T):
        self.__item = item
        print(f'Stored: {self.__item}')

    def retrieve(self) -> T:
        if self.__item is None:
            print('Nothing stored yet')
        else:
            print(f'Retrieved: {self.__item}')
            return self.__item


int_storage = Storage()
int_storage.store(42)
int_storage.retrieve()

str_storage = Storage()
str_storage.store('Akyaa')
str_storage.retrieve()

list_storage = Storage()
list_storage.store([1, 2, 3, 4, 5])
list_storage.retrieve()
```

# Case Study Activity 2: Warehouse Inventory System

```python
from typing import TypeVar, Generic, List

T = TypeVar('T')

class Inventory(Generic[T]):
    def __init__(self):
        self.__items: List[T] = []

    def store(self, item: T):
        self.__items.append(item)
        print(f'Stored: {item} (Type: {type(item).__name__})')

    def retrieve(self, index: int) -> T:
        if index < len(self.__items):
            print(f'Retrieved: {self.__items[index]}')
            return self.__items[index]
        else:
            print('Item not found')

    def display(self):
        for item in self.__items:
            print(f'Value: {item} | Type: {type(item).__name__}')


str_inventory = Inventory()
str_inventory.store('Laptop')
str_inventory.store('Phone')
str_inventory.display()

print()

int_inventory = Inventory()
int_inventory.store(100)
int_inventory.store(250)
int_inventory.display()

print()

float_inventory = Inventory()
float_inventory.store(49.99)
float_inventory.store(199.99)
float_inventory.display()

print()

list_inventory = Inventory()
list_inventory.store(['SN001', 'SN002', 'SN003'])
list_inventory.store(['SN004', 'SN005'])
list_inventory.display()
```
